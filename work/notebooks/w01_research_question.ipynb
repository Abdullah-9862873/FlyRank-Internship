## 1. My lane (or freestyle) and why

> I pick the **Refresh / Content Opportunity Scoring** lane. The idea: recommend which content pages to refresh (rework) first. Over half of the pages in the starter data are already measured as declining in traffic, so ranking which one to refresh first adds real value. It also matches the trend and impressions questions I already explored, so I can reuse what I have learned.

```python
import os
import pandas as pd

def locate_csv():
    roots = ['/content', './', '../', '../../data/raw', '/']
    for r in roots:
        for dirpath, _, files in os.walk(r):
            if 'content_refresh_anonymized.csv' in files:
                return os.path.join(dirpath, 'content_refresh_anonymized.csv')
    return None

path = locate_csv()
if path is None:
    raise FileNotFoundError('CSV not found anywhere. Upload content_refresh_anonymized.csv to Colab.')
print('Using:', path)

df = pd.read_csv(path)

declining_share = (df['trend_direction'] == 'down').mean()
declining_pages = (df['trend_direction'] == 'down').sum()
total_imp = df['impressions_90d'].sum()

print('Rows:', len(df))
print('Declining share:', round(declining_share * 100, 1), '%')
print('Declining pages:', declining_pages)
print('Total 90-day impressions:', total_imp)
```

## 2. The question: decision, action, cost of a wrong call

**Question:** Which content pages should a publisher refresh (rework) first?

**Decision it improves:** Which page to refresh next.

**Who acts on it:** An editor who picks the highest-ranked page from the list and reworks it.

**Cost of a wrong recommendation:** Two real costs. First, effort spent reworking a page that was already healthy. Second, missing a page that keeps losing traffic, so it falls for another cycle. Both waste time and lost impressions.

```python
# This cell backs the decision framing with a base-rate check.
import os
import pandas as pd

def locate_csv():
    roots = ['/content', './', '../', '../../data/raw', '/']
    for r in roots:
        for dirpath, _, files in os.walk(r):
            if 'content_refresh_anonymized.csv' in files:
                return os.path.join(dirpath, 'content_refresh_anonymized.csv')
    return None

path = locate_csv()
if path is None:
    raise FileNotFoundError('CSV not found anywhere. Upload content_refresh_anonymized.csv to Colab.')
print('Using:', path)

df = pd.read_csv(path)

base_rate = (df['trend_direction'] == 'down').mean()
print('Base rate of declining pages:', round(base_rate * 100, 1), '%')
print('A random pick is close to a coin flip, so ranking has room to help.')
```

## 4. Careful words: what I can and can't claim

**What I can never claim:** That refreshing a page causes traffic to recover (that would need causal checks), a specific traffic number, or that I can predict Google's changes. I only rank which pages look most likely to be worth a refresh look first.
